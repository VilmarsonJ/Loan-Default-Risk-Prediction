- **Type d’emploi :** Les défauts de remboursement sont légèrement plus élevés chez les **employés à temps partiel (3 %)** et les **travailleurs indépendants (2,9 %)**, tandis que les **employés à temps plein présentent le risque le plus faible (2,4 %)**. Cela suggère que la **stabilité professionnelle** influence fortement la capacité de remboursement.  

- **Statut marital :** Les emprunteurs **divorcés (4,2 %)** et **célibataires (4 %)** affichent un risque de défaut plus élevé comparé aux **mariés (3,5 %)**, indiquant que le **soutien familial et financier** peut jouer un rôle dans le comportement de remboursement.  

- **Niveau d’éducation :** Les défauts sont légèrement plus fréquents chez ceux avec un **baccalauréat (3,1 %) ou un diplôme secondaire (3,2 %)**, tandis que les **masters (2,7 %) et doctorats (2,6 %)** présentent un risque plus faible. Cela reflète l’influence du **capital humain et des compétences financières** sur la capacité à gérer les obligations de prêt.  

- **Autres variables catégorielles :** Les informations sur la possession d’un **prêt hypothécaire**, la présence de **dépendants**, la **finalité du prêt** et la présence d’un **cosignataire** permettent également d’identifier les segments à risque et guider les **décisions stratégiques de crédit**.

```python
# -----------------------------
# 2️⃣ Analyse des Variables Catégorielles
# -----------------------------
# S'assurer que le dossier Images existe
if not os.path.exists("Images"):
    os.makedirs("Images")

# Créer des labels explicites pour la variable cible
loan['Default_label_fr'] = loan['Default'].map({
    0: 'Remboursé à Temps',
    1: 'Défaut de remboursement'
})

# Liste des variables catégorielles à explorer
categorical_cols = ['Education', 'EmploymentType', 'MaritalStatus', 
                    'HasMortgage', 'HasDependents', 'LoanPurpose', 'HasCoSigner']

# Boucle sur chaque variable catégorielle
for col in categorical_cols:
    plt.figure(figsize=(10, 6))
    
    # Création du graphique
    ax = sns.countplot(
        data=loan, 
        x=col, 
        hue='Default_label_fr',
        palette=["#2E86AB", "#D9534F"],  # Bleu = sain, Rouge = risqué
        edgecolor='black'
    )
    
    # Ajouter les pourcentages sur chaque barre
    total = len(loan)
    for p in ax.patches:
        percentage = f"{100 * p.get_height()/total:.1f}%"
        ax.annotate(
            percentage, 
            (p.get_x() + p.get_width() / 2., p.get_height()),
            ha='center', va='bottom', fontsize=11, fontweight='bold', color='black'
        )

    # 🔹 Amélioration du titre et des axes
    plt.title(f"Distribution de {col} selon le Statut du Prêt", fontsize=18, fontweight='bold', pad=15)
    plt.xlabel(col, fontsize=14, fontweight='bold')
    plt.ylabel("Nombre de Prêts", fontsize=14, fontweight='bold')

    # 🔹 Mettre les labels de l’axe X en majuscules et en gras
    ax.set_xticklabels(
        [label.get_text().upper() for label in ax.get_xticklabels()],
        fontsize=12, fontweight='bold'
    )

    # 🔹 Légende claire et concise, avec texte en gras
    legend = plt.legend(
        title='Statut du Prêt',
        title_fontsize=13,
        fontsize=13,
        loc='upper right',
        frameon=True
    )
    plt.setp(legend.get_texts(), fontweight='bold')

    # 🔹 Ajustement final
    plt.xticks(rotation=45)
    plt.tight_layout()

    # 🔹 Sauvegarder chaque graphique automatiquement dans le dossier Images
    filename = f"Images/distribution_{col.lower()}_fr.png"
    plt.savefig(filename, dpi=300, bbox_inches="tight", facecolor="white")

    plt.show()
```

### EDA – Analyse Orientée Business
----
L’Analyse Exploratoire des Données (EDA) offre une vue complète sur le **comportement de remboursement des prêts** et met en évidence les principaux **facteurs de risque** selon les caractéristiques démographiques, professionnelles et financières des emprunteurs. Les observations ci-dessous sont présentées dans un contexte business, avec des recommandations exploitables pour les parties prenantes, y compris les banques en Haïti ou dans d’autres marchés émergents.

1. **Performance Globale des Prêts (Diagramme circulaire)**  
   Le diagramme circulaire montre qu’environ **88% des emprunteurs remboursent leurs prêts à temps**, tandis que **12% font défaut**. Même si la majorité des clients sont fiables, ce **segment minoritaire mais critique** représente un **risque financier important** pour les banques. Identifier proactivement ces emprunteurs peut **prévenir des pertes** et stabiliser le portefeuille de prêts.

2. **Type d’Emploi**  
   Les emprunteurs **indépendants ou à temps partiel** présentent un taux de défaut légèrement supérieur à celui des **employés à temps plein**, indiquant que la **stabilité des revenus** est un facteur clé pour le remboursement. Les banques pourraient en tenir compte lors de l’approbation des prêts ou pour ajuster les taux d’intérêt selon le profil de risque.

3. **Statut Matrimonial**  
   L’analyse montre que les **divorcés et célibataires** sont plus susceptibles de faire défaut que les personnes **mariées**. Cela suggère que le **soutien familial et la stabilité du foyer** influencent positivement le remboursement. Les stratégies de prêt peuvent intégrer ce facteur pour concevoir des **interventions ciblées**.

4. **Niveau d’Éducation**  
   Les emprunteurs ayant un niveau d’études **supérieur (Master, PhD)** ont tendance à moins faire défaut que ceux ayant un niveau **Bachelor ou secondaire**, soulignant l’importance de la **littératie financière et du potentiel de revenu** dans la gestion du risque. L’éducation peut servir de **proxy pour la stabilité financière** dans les modèles de scoring.

5. **Autres Facteurs Catégoriels**  
   Des variables comme **propriétaire d’une hypothèque, nombre de personnes à charge, objectif du prêt ou présence d’un co-emprunteur** fournissent également des informations précieuses. Par exemple, un co-emprunteur ou une expérience hypothécaire antérieure est généralement associé à une **meilleure fiabilité de remboursement**, mettant en avant l’importance des **antécédents financiers et de la garantie sociale**.

**Principaux enseignements pour les décideurs :**  
- Bien que le taux de défaut soit relativement faible, l’**impact financier des défauts reste significatif**, et une surveillance proactive est essentielle.  
- Le **type d’emploi, le statut matrimonial et le niveau d’éducation** sont des prédicteurs pertinents pour évaluer le risque et guider des **politiques de crédit ciblées**.  
- Ces insights permettent aux banques de concevoir des **stratégies de prêt adaptées au risque**, de développer des **mesures préventives pour les emprunteurs à risque** et d’améliorer la **stabilité globale du portefeuille**.

> L’EDA fournit une base solide pour le **feature engineering et le modeling prédictif**. Comprendre ces patterns permet aux institutions financières de mettre en place des **stratégies basées sur les données** pour réduire les défauts tout en favorisant un prêt responsable.
