- #### **Model Loading & Deployment**

Une fois les modèles sauvegardés, il est temps de les **recharger (load)** afin de les **déployer dans un environnement applicatif**.  
Grâce à la bibliothèque `joblib`, cette étape se fait en une ligne de code : elle recharge les poids et paramètres optimisés sans réexécuter tout l’entraînement.

Ces modèles seront utilisés pour :
- **Faire de nouvelles prédictions** sur des dossiers de prêt récents ;

- **Alimenter le Dashboard**, qui permettra aux décideurs de manipuler les résultats sans écrire de code.

> En pratique, cela signifie qu’une **coopérative d’épargne à Jacmel** ou une **banque à Cap-Haïtien** peut charger le modèle sauvegardé et **analyser instantanément le risque de défaut** de nouveaux clients, en profitant de la puissance du machine learning sans expertise technique.

Cette étape marque la transition entre la **modélisation** et la **prise de décision stratégique**, où la data science devient un **levier opérationnel** au service du développement financier haïtien.

---


```python
import joblib
import pandas as pd
from joblib import load

# Charger le preprocessor sauvegardé
preprocessor = joblib.load("../Models/preprocessor_final.pkl")

# Charger le dataset clean
loan = pd.read_csv("../Data/processed/Loan_clean.csv")

# Supprimer les colonnes non pertinentes (comme dans l'entraînement)
drop_cols = ['LoanID', 'Default', 'Default_label_fr' ]
X = loan.drop(columns=drop_cols)

# Appliquer le preprocessing (exactement comme pour l'entraînement)
X_processed = preprocessor.transform(X)

# Charger les modèles
# logit_model = joblib.load("../Models/logit_optimized.joblib")
rf_model = joblib.load("../Models/model_final.joblib")

# Faire les prédictions

loan['RF_Pred'] = rf_model.predict(X_processed)
loan['RF_Prob'] = rf_model.predict_proba(X_processed)[:, 1]

# Sauvegarder les prédictions
loan.to_csv("../Data/processed/predictions.csv", index=False)
print("✅ Prédictions générées et sauvegardées avec succès !")
```

## **Limites et Points de Vigilance**

Bien que les modèles développés offrent une valeur significative pour anticiper le risque de défaut, plusieurs **limites et considérations business** doivent être prises en compte avant de déployer les recommandations à grande échelle :


#### **1. Qualité et représentativité des données**
- Les données utilisées reflètent un **échantillon historique limité** d’emprunteurs.  
- Certaines variables critiques (revenu réel, fluctuations économiques locales, endettement informel) peuvent être **sous-représentées ou manquantes**, ce qui peut influencer la fiabilité des prédictions.  
- **Impact business :** Les décisions basées uniquement sur ce modèle pourraient **sous-estimer le risque** pour des segments non représentés dans les données.


#### **2. Modèle basé sur des tendances passées**
- Les modèles apprennent des comportements historiques des emprunteurs.  
- Les changements rapides dans l’économie, la législation financière ou le comportement des consommateurs peuvent **altérer la performance prédictive**.  
- **Impact business :** Une surveillance régulière est essentielle pour éviter que le modèle devienne obsolète et **ne reflète plus la réalité du marché**.


#### **3. Limites d’interprétation et de causalité**
- La **Random Forest** et autres modèles complexes détectent des patterns et corrélations, mais **ne prouvent pas de causalité**.  
- Certaines décisions (refuser un prêt, ajuster un taux) doivent combiner les résultats du modèle avec **l’expertise humaine et la connaissance du marché**.  
- **Impact business :** La confiance excessive dans les prédictions automatiques peut conduire à des **erreurs stratégiques coûteuses** si elle n’est pas contextualisée.

#### **4. Risque d’effets discriminatoires ou biaisés**
- Certaines caractéristiques socio-démographiques peuvent **induire un biais involontaire** si elles sont fortement corrélées à la probabilité de défaut.  
- Il est crucial de vérifier la conformité avec les **règles de crédit équitable** et d’éviter tout traitement discriminatoire.  
- **Impact business :** Une mauvaise gestion de ce risque peut affecter la **réputation de l’institution** et entraîner des conséquences réglementaires ou légales.

#### **5. Limites techniques et scalabilité**
- Les modèles nécessitent une **infrastructure informatique adaptée** pour des mises à jour et des prédictions en temps réel.  
- Les volumes de données futurs, l’intégration multi-branches ou la connexion à d’autres systèmes peuvent révéler des **limitations de performance ou de rapidité**.
